```python
import os
import zipfile
from typing import List

import pandas as pd


def unzip_dataset(zip_path: str, extract_path: str) -> None:
    """
    Unzips the dataset to the given directory.
    """
    with zipfile.ZipFile(zip_path, "r") as zip_ref:
        zip_ref.extractall(extract_path)


def get_csv_files(folder_path: str) -> List[str]:
    """
    Returns all CSV file paths inside a folder.
    """
    csv_files = []

    for root, _, files in os.walk(folder_path):
        for file in files:
            if file.lower().endswith(".csv"):
                csv_files.append(os.path.join(root, file))

    if not csv_files:
        raise FileNotFoundError("No CSV files found")

    return csv_files


def read_csv_file_safe(file_path: str) -> pd.DataFrame:
    """
    Reads a CSV file with encoding fallback.
    """
    try:
        return pd.read_csv(file_path, encoding="utf-8")
    except UnicodeDecodeError:
        return pd.read_csv(file_path, encoding="latin1")


def read_csv_files(csv_files: List[str]) -> List[pd.DataFrame]:
    """
    Reads all CSV files safely.
    """
    dataframes = []

    for file in csv_files:
        df = read_csv_file_safe(file)
        dataframes.append(df)

    return dataframes


def merge_dataframes(dataframes: List[pd.DataFrame]) -> pd.DataFrame:
    """
    Merge multiple DataFrames into one.
    """
    return pd.concat(dataframes, ignore_index=True)


def save_dataframe(df: pd.DataFrame, output_path: str) -> None:
    """
    Save DataFrame to CSV.
    """
    df.to_csv(output_path, index=False)


def main() -> None:
    zip_file_path = "/content/RESOURCE.zip"
    extract_folder = "/content/extracted_data"
    output_file = "/content/merged_dataset.csv"

    unzip_dataset(zip_file_path, extract_folder)

    csv_files = get_csv_files(extract_folder)
    print(f"📂 CSV files found: {len(csv_files)}")

    dataframes = read_csv_files(csv_files)
    merged_df = merge_dataframes(dataframes)

    save_dataframe(merged_df, output_file)

    print("✅ Dataset merged successfully!")
    print(f"📄 Output file: {output_file}")
    print(f"📊 Final shape: {merged_df.shape}")


if __name__ == "__main__":
    main()
```

```python
import pandas as pd
from IPython.display import display


def load_dataframe(file_path: str) -> pd.DataFrame:
    """
    Load merged CSV into DataFrame.
    """
    return pd.read_csv(file_path)


def basic_info(df: pd.DataFrame) -> None:
    """
    Display basic information about the dataset.
    """
    print("🔹 Dataset Shape (Rows, Columns):")
    print(df.shape)

    print("\n🔹 Column Names:")
    print(df.columns.tolist())

    print("\n🔹 Data Types:")
    print(df.dtypes)


def missing_value_analysis(df: pd.DataFrame) -> None:
    """
    Display missing value count per column.
    """
    missing_values = df.isnull().sum()
    result = missing_values[missing_values > 0]

    print("\n🔹 Missing Values Per Column:")
    if result.empty:
        print("✅ No missing values found.")
    else:
        print(result)


def statistical_summary(df: pd.DataFrame) -> None:
    """
    Display statistical summary for numerical columns.
    """
    print("\n🔹 Statistical Summary:")
    display(df.describe())




def data_quality_checks(df: pd.DataFrame) -> None:
    """
    Perform basic data quality checks.
    """
    print("\n🔹 DATA QUALITY ANALYSIS")


    duplicate_count = df.duplicated().sum()
    print(f"🔸 Duplicate records: {duplicate_count}")


    numeric_columns = df.select_dtypes(include="number").columns
    negative_values = (df[numeric_columns] < 0).sum().sum()
    print(f"🔸 Negative values found: {negative_values}")


    zero_values = (df[numeric_columns] == 0).sum()
    dominant_zero_columns = zero_values[zero_values > 0]

    print("\n🔸 Columns with zero values:")
    print(dominant_zero_columns.head())


    if (
        "Main Workers - Total - Persons" in df.columns
        and "Main Workers - Total - Males" in df.columns
        and "Main Workers - Total - Females" in df.columns
    ):
        inconsistent_rows = df[
            df["Main Workers - Total - Persons"]
            != df["Main Workers - Total - Males"]
            + df["Main Workers - Total - Females"]
        ]

        print(
            f"\n🔸 Inconsistent gender totals (Main Workers): "
            f"{len(inconsistent_rows)} rows"
        )


def sample_records(df: pd.DataFrame) -> None:
    """
    Display sample rows from dataset.
    """
    print("\n🔹 Sample Records:")
    display(df.head(10))


def main() -> None:
    file_path = "/content/merged_dataset.csv"

    df = load_dataframe(file_path)

    print("✅ Data Exploration Started")

    basic_info(df)
    missing_value_analysis(df)
    statistical_summary(df)
    data_quality_checks(df)
    sample_records(df)

    print("\n✅ Data Exploration Completed")


if __name__ == "__main__":
    main()
```

```python
import pandas as pd



def load_dataframe(file_path: str) -> pd.DataFrame:
    """
    Load CSV file into DataFrame.
    """
    return pd.read_csv(file_path)


def clean_column_names(df: pd.DataFrame) -> pd.DataFrame:
    """
    Standardize column names to snake_case.
    """
    df.columns = (
        df.columns.str.strip()
        .str.lower()
        .str.replace(r"[^\w\s]", " ", regex=True)
        .str.replace(r"\s+", "_", regex=True)
    )
    return df


def clean_string_columns(df: pd.DataFrame) -> pd.DataFrame:
    """
    Remove unwanted characters from string columns.
    """
    string_columns = df.select_dtypes(include="object").columns

    for col in string_columns:
        df[col] = (
            df[col]
            .astype(str)
            .str.replace("`", "", regex=False)
            .str.strip()
        )

    return df


def create_nlp_text_column(df: pd.DataFrame) -> pd.DataFrame:
    """
    Create a cleaned text column for NLP from industry name.
    """
    df["industry_text"] = (
        df["nic_name"]
        .str.lower()
        .str.replace(r"[^a-z\s]", "", regex=True)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

    return df


def data_cleaning_pipeline(df: pd.DataFrame) -> pd.DataFrame:
    """
    Complete data cleaning pipeline.
    """
    df = clean_column_names(df)
    df = clean_string_columns(df)
    df = create_nlp_text_column(df)
    return df


def save_cleaned_data(df: pd.DataFrame, output_path: str) -> None:
    """
    Save cleaned DataFrame.
    """
    df.to_csv(output_path, index=False)


def main() -> None:
    input_file = "/content/merged_dataset.csv"
    output_file = "/content/cleaned_dataset.csv"

    df = load_dataframe(input_file)
    cleaned_df = data_cleaning_pipeline(df)
    save_cleaned_data(cleaned_df, output_file)

    print("✅ Data Cleaning Completed Successfully")
    print(f"📄 Cleaned dataset saved at: {output_file}")
    print(f"📊 Shape after cleaning: {cleaned_df.shape}")
    print("\n🔍 Cleaned Columns:")
    print(cleaned_df.columns.tolist())


if __name__ == "__main__":
    main()
```

```python
import pandas as pd


def load_cleaned_data(file_path: str) -> pd.DataFrame:
    """
    Load cleaned dataset.
    """
    return pd.read_csv(file_path)


def create_worker_totals(df: pd.DataFrame) -> pd.DataFrame:
    """
    Create total worker-related features.
    """
    df["total_workers"] = (
        df["main_workers_total_persons"]
        + df["marginal_workers_total_persons"]
    )

    df["total_male_workers"] = (
        df["main_workers_total_males"]
        + df["marginal_workers_total_males"]
    )

    df["total_female_workers"] = (
        df["main_workers_total_females"]
        + df["marginal_workers_total_females"]
    )

    return df


def create_area_based_features(df: pd.DataFrame) -> pd.DataFrame:
    """
    Create rural and urban worker totals.
    """
    df["total_rural_workers"] = (
        df["main_workers_rural_persons"]
        + df["marginal_workers_rural_persons"]
    )

    df["total_urban_workers"] = (
        df["main_workers_urban_persons"]
        + df["marginal_workers_urban_persons"]
    )

    return df


def create_worker_type_feature(df: pd.DataFrame) -> pd.DataFrame:
    """
    Identify dominant worker type.
    """
    df["dominant_worker_type"] = df.apply(
        lambda row: "Main"
        if row["main_workers_total_persons"]
        >= row["marginal_workers_total_persons"]
        else "Marginal",
        axis=1
    )

    return df


def create_nlp_features(df: pd.DataFrame) -> pd.DataFrame:
    """
    Create NLP-related features.
    """
    df["industry_word_count"] = df["industry_text"].apply(
        lambda x: len(str(x).split())
    )

    return df


def feature_engineering_pipeline(df: pd.DataFrame) -> pd.DataFrame:
    """
    Complete feature engineering pipeline.
    """
    df = create_worker_totals(df)
    df = create_area_based_features(df)
    df = create_worker_type_feature(df)
    df = create_nlp_features(df)

    return df


def save_feature_engineered_data(df: pd.DataFrame, output_path: str) -> None:
    """
    Save feature engineered dataset.
    """
    df.to_csv(output_path, index=False)


def main() -> None:
    input_file = "/content/cleaned_dataset.csv"
    output_file = "/content/feature_engineered_dataset.csv"

    df = load_cleaned_data(input_file)
    fe_df = feature_engineering_pipeline(df)

    save_feature_engineered_data(fe_df, output_file)

    print("✅ Feature Engineering Completed")
    print(f"📄 Saved at: {output_file}")
    print(f"📊 Final shape: {fe_df.shape}")
    print("\n🔍 New Features Added:")
    print(
        [
            "total_workers",
            "total_male_workers",
            "total_female_workers",
            "total_rural_workers",
            "total_urban_workers",
            "dominant_worker_type",
            "industry_word_count",
        ]
    )


if __name__ == "__main__":
    main()
```

```python
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split


def load_feature_engineered_data(file_path: str) -> pd.DataFrame:
    """
    Load feature engineered dataset.
    """
    return pd.read_csv(file_path)


def create_business_category(industry_text: str) -> str:
    """
    Rule-based NLP mapping for business category.
    """
    text = str(industry_text).lower()

    if any(word in text for word in ["retail", "shop", "trade"]):
        return "Retail"
    if any(word in text for word in ["poultry", "chicken", "livestock"]):
        return "Poultry"
    if any(word in text for word in ["agriculture", "crop", "farming", "cultivation"]):
        return "Agriculture"
    if any(word in text for word in ["manufacture", "factory", "production"]):
        return "Manufacturing"
    if any(word in text for word in ["construction", "building", "infrastructure"]):
        return "Construction"

    return "Others"


def prepare_target_variable(df: pd.DataFrame) -> pd.DataFrame:
    """
    Create target variable for model.
    """
    df["business_category"] = df["industry_text"].apply(
        create_business_category
    )
    return df


def split_data(df: pd.DataFrame):
    """
    Split dataset into train and test sets.
    """
    x = df["industry_text"]
    y = df["business_category"]

    return train_test_split(
        x, y, test_size=0.2, random_state=42, stratify=y
    )


def train_model(x_train, y_train):
    """
    Train TF-IDF + Logistic Regression model.
    """
    vectorizer = TfidfVectorizer(stop_words="english", max_features=5000)
    x_train_tfidf = vectorizer.fit_transform(x_train)

    model = LogisticRegression(max_iter=1000)
    model.fit(x_train_tfidf, y_train)

    return model, vectorizer


def evaluate_model(model, vectorizer, x_test, y_test) -> None:
    """
    Evaluate trained model.
    """
    x_test_tfidf = vectorizer.transform(x_test)
    predictions = model.predict(x_test_tfidf)

    accuracy = accuracy_score(y_test, predictions)

    print("✅ MODEL EVALUATION RESULTS")
    print(f"🎯 Accuracy: {accuracy:.2f}")
    print("\n📊 Classification Report:")
    print(classification_report(y_test, predictions))


def main() -> None:
    file_path = "/content/feature_engineered_dataset.csv"

    df = load_feature_engineered_data(file_path)
    df = prepare_target_variable(df)

    x_train, x_test, y_train, y_test = split_data(df)

    model, vectorizer = train_model(x_train, y_train)

    evaluate_model(model, vectorizer, x_test, y_test)


if __name__ == "__main__":
    main()
```

```python
%%writefile app.py
import pandas as pd
import plotly.express as px
import streamlit as st


@st.cache_data
def load_data() -> pd.DataFrame:
    """
    Load feature engineered dataset.
    """
    return pd.read_csv("feature_engineered_dataset.csv")


def business_category_mapping(industry_text: str) -> str:
    """
    Rule-based NLP mapping to business categories.
    """
    text = str(industry_text).lower()

    if any(word in text for word in ["retail", "trade", "shop"]):
        return "Retail"
    if any(word in text for word in ["poultry", "chicken"]):
        return "Poultry"
    if any(word in text for word in ["agriculture", "crop", "farming"]):
        return "Agriculture"
    if any(word in text for word in ["manufacture", "production", "factory"]):
        return "Manufacturing"
    if any(word in text for word in ["construction", "building"]):
        return "Construction"

    return "Others"


def prepare_dashboard_data(df: pd.DataFrame) -> pd.DataFrame:
    """
    Prepare business category column.
    """

    df = df[df["india_states"].str.startswith("STATE")].copy()

    df["business_category"] = df["industry_text"].apply(
        business_category_mapping
    )
    return df




def plot_workers_by_category(df: pd.DataFrame) -> None:
    """
    Bar chart: Total workers by business category.
    """

    grouped = df.groupby("business_category")["total_workers"].sum().reset_index()

    fig = px.bar(
        grouped,
        x="business_category",
        y="total_workers",
        title="Total Workers by Business Category"
    )
    st.plotly_chart(fig, width="stretch")


def plot_workers_by_state(df: pd.DataFrame) -> None:
    """
    Bar chart: Top 10 states by worker population.
    """


    grouped = (
        df.groupby("india_states")["total_workers"]
        .sum()
        .reset_index()
        .sort_values("total_workers", ascending=False)
        .head(10)
    )

    fig = px.bar(
        grouped,
        x="india_states",
        y="total_workers",
        title="Top 10 States by Worker Population"
    )
    st.plotly_chart(fig, width="stretch")




def plot_rural_urban_split(df: pd.DataFrame) -> None:
    """
    Pie chart: Rural vs Urban worker distribution.
    """

    data = pd.DataFrame({
        "Area": ["Rural", "Urban"],
        "Workers": [
            df["total_rural_workers"].sum(),
            df["total_urban_workers"].sum()
        ]
    })

    fig = px.pie(
        data,
        names="Area",
        values="Workers",
        title="Rural vs Urban Worker Distribution"
    )
    st.plotly_chart(fig, width="stretch")


def plot_gender_distribution(df: pd.DataFrame) -> None:
    """
    Bar chart: Male vs Female worker distribution.
    """

    data = pd.DataFrame({
        "Gender": ["Male", "Female"],
        "Workers": [
            df["total_male_workers"].sum(),
            df["total_female_workers"].sum()
        ]
    })

    fig = px.bar(
        data,
        x="Gender",
        y="Workers",
        title="Gender-wise Worker Distribution"
    )
    st.plotly_chart(fig, width="stretch")


def plot_female_participation_by_industry(df: pd.DataFrame) -> None:
    """
    Bar chart: Female workforce by business category.
    """

    grouped = (
        df.groupby("business_category")["total_female_workers"]
        .sum()
        .reset_index()
    )

    fig = px.bar(
        grouped,
        x="business_category",
        y="total_female_workers",
        title="Female Workforce by Industry"
    )
    st.plotly_chart(fig, width="stretch")


def display_facts(df: pd.DataFrame) -> None:
    """
    Display key facts and figures about the workforce.
    """

    st.subheader("📌 Key Facts & Figures")

    st.write(f"• **Total Workforce:** {int(df['total_workers'].sum()):,}")
    st.write(
        f"• **Dominant Industry:** "
        f"{df.groupby('business_category')['total_workers'].sum().drop('Others').idxmax()}"
    )
    st.write(
        f"• **Highest Employment State:** "
        f"{df.groupby('india_states')['total_workers'].sum().idxmax()}"
    )
    st.write(
        f"• **Total Rural Workers:** {int(df['total_rural_workers'].sum()):,}"
    )
    st.write(
        f"• **Total Urban Workers:** {int(df['total_urban_workers'].sum()):,}"
    )



def main() -> None:
    st.set_page_config(
        page_title="Industrial Workforce Dashboard",
        layout="wide"
    )

    st.title("📊 Industrial Human Resource Geo-Visualization")

    df = load_data()
    df = prepare_dashboard_data(df)

    section = st.selectbox(
        "Select Analysis View",
        (
            "Industry-wise Analysis",
            "Geography-wise Analysis",
            "Rural vs Urban Distribution",
            "Gender Analysis",
            "Key Facts & Insights"
        )
    )

    if section == "Industry-wise Analysis":
        plot_workers_by_category(df)
        plot_female_participation_by_industry(df)

    elif section == "Geography-wise Analysis":
        plot_workers_by_state(df)


    elif section == "Rural vs Urban Distribution":
        plot_rural_urban_split(df)

    elif section == "Gender Analysis":
        plot_gender_distribution(df)

    elif section == "Key Facts & Insights":
        display_facts(df)


if __name__ == "__main__":
    main()
```
